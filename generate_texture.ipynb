```python
from mtex import *
```

# Generate a texture

```python
# PF plotting conventions
plottingConvention.default('y↑→x')
cs = crystalFrame('m-3m', [4.05, 4.05, 4.05], mineral='Aluminium', color='blue')
ss = specimenFrame('triclinic')
psi = SO3DeLaValleePoussinKernel(halfwidth=7*degree);
hkil = [Miller(0, 0, 1, cs), Miller(1,1,0, cs), Miller(1,1,1, cs)]
```

## From one euler angle

```python

odf = unimodalODF(orientation.byEuler(0*degree, 0*degree, 0*degree, cs), halfwidth=15*degree)
plotPF(odf, hkil, contourf=True, complete=True, upper=True, minmax=True)
setColorRange('equal')
mtexColorbar(location='eastoutside', label='mrd', multiple=False)
```

## From one quaternion

```python

odf = unimodalODF(orientation(quaternion(1, 0, 0, 0), cs), halfwidth=15*degree)
plotPF(odf, hkil, contourf=True, complete=True, upper=True, minmax=True)
setColorRange('equal')
mtexColorbar(location='eastoutside', label='mrd', multiple=False)
```
